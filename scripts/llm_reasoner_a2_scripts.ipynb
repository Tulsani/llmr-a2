### b) *part*

##### Mean Calc

```python
import subprocess
import pandas as pd
import re

sizes = ["small", "medium", "large", "xl", "2.7B"]
modes = ["forward", "backward", "forward_backward"]

results = []

for size in sizes:
    for mode in modes:
        print(f"\nRunning {size} | {mode}")

        cmd = f"""
        uv run python ./student/benchmarking_script.py \
            --size {size} \
            --context_length 512 \
            --mode {mode} \
            --warmup_steps 5 \
            --n_steps 10
        """

        output = subprocess.check_output(cmd, shell=True, text=True)
        print(output)

        # extract time
        match = re.search(r"Mean\s*:\s*([\d\.]+)\s*ms", output)
        if match:
            time_ms = float(match.group(1))
            results.append({
                "size": size,
                "mode": mode,
                "mean_ms": time_ms
            })

df = pd.DataFrame(results)
df
```

### 2. profiling script a)

```python
import os
import subprocess
import pandas as pd

sizes = ["small", "medium", "large", "xl", "2.7B"]
contexts = [128, 256, 512, 1024]
modes = ["forward", "backward", "forward_backward"]

os.makedirs("nsys_profiles", exist_ok=True)
os.makedirs("nsys_stats", exist_ok=True)

for size in sizes:
    for ctx in contexts:
        for mode in modes:
            tag = f"{size}_ctx{ctx}_{mode}"
            out_base = f"nsys_profiles/{tag}"

            cmd = f"""
            nsys profile \
              --force-overwrite true \
              -t cuda,nvtx,osrt \
              -o {out_base} \
              uv run python ./student/nsys_profile.py \
                --size "{size}" \
                --context_length {ctx} \
                --mode {mode} \
                --warmup_steps 5 \
                --n_steps 10
            """

            print(f"\nRunning {tag}")
            try:
                subprocess.run(cmd, shell=True, check=True)

                stats_cmd = f"""
                nsys stats \
                  --report nvtx_sum \
                  --format csv \
                  {out_base}.nsys-rep > nsys_stats/{tag}_nvtx.csv
                """
                subprocess.run(stats_cmd, shell=True, check=True)

            except subprocess.CalledProcessError:
                print(f"OOM or failed for {tag}")
```

## Part 3: Mixed Precision Profiling

### Part c)

```python
import subprocess
import pandas as pd
import re

sizes = ["small", "medium", "large", "xl", "2.7B"]
modes = ["forward", "backward", "forward_backward"]

results = []

for size in sizes:
    for mode in modes:
        for mixed in [False, True]:
            precision = "bf16_mixed" if mixed else "fp32"

            cmd = f"""
            uv run python ./student/mixed_precision_profiling.py \
                --size "{size}" \
                --context_length 512 \
                --mode {mode} \
                --warmup_steps 5 \
                --n_steps 10 \
                {"--mixed_precision" if mixed else ""}
            """

            print(f"\nRunning size={size} | mode={mode} | precision={precision}")

            try:
                output = subprocess.check_output(cmd, shell=True, text=True, stderr=subprocess.STDOUT)
                print(output)

                mean_match = re.search(r"Mean\s*:\s*([\d\.]+)\s*ms", output)
                std_match = re.search(r"Std\s*:\s*([\d\.]+)\s*ms", output)
                param_match = re.search(r"Parameters:\s*([\d\.]+)M", output)

                results.append({
                    "size": size,
                    "mode": mode,
                    "precision": precision,
                    "params_m": float(param_match.group(1)) if param_match else None,
                    "mean_ms": float(mean_match.group(1)) if mean_match else None,
                    "std_ms": float(std_match.group(1)) if std_match else None,
                    "status": "ok",
                })

            except subprocess.CalledProcessError as e:
                print(e.output)
                results.append({
                    "size": size,
                    "mode": mode,
                    "precision": precision,
                    "params_m": None,
                    "mean_ms": None,
                    "std_ms": None,
                    "status": "failed/OOM",
                })

df = pd.DataFrame(results)
df
```

## Part 4: Memory Profiling

#### Full precision 2.7b profiling

```python
import os
import subprocess
import pandas as pd
import re

size = "2.7B"
contexts = [128, 256, 512]
modes = ["forward", "forward_backward"]   # inference only vs full training step

os.makedirs("memory_snapshots", exist_ok=True)

results = []

for ctx in contexts:
    for mode in modes:
        snapshot_file = f"memory_snapshots/{size}_ctx{ctx}_{mode}.pickle"

        cmd = f"""
        uv run python ./student/memory_profiling.py \
            --size "{size}" \
            --context_length {ctx} \
            --mode {mode} \
            --warmup_steps 5 \
            --profile_memory \
            --snapshot_file "{snapshot_file}"
        """

        print(f"\nRunning size={size} | ctx={ctx} | mode={mode}")

        try:
            output = subprocess.check_output(
                cmd,
                shell=True,
                text=True,
                stderr=subprocess.STDOUT
            )
            print(output)

            peak_match = re.search(r"Peak memory allocated:\s*([\d\.]+)\s*MB", output)
            param_match = re.search(r"Parameters:\s*([\d\.]+)M", output)

            results.append({
                "size": size,
                "context_length": ctx,
                "mode": mode,
                "params_m": float(param_match.group(1)) if param_match else None,
                "peak_memory_mb": float(peak_match.group(1)) if peak_match else None,
                "snapshot_file": snapshot_file,
                "status": "ok",
            })

        except subprocess.CalledProcessError as e:
            print(e.output)
            results.append({
                "size": size,
                "context_length": ctx,
                "mode": mode,
                "params_m": None,
                "peak_memory_mb": None,
                "snapshot_file": snapshot_file,
                "status": "failed/OOM",
            })

df = pd.DataFrame(results)
df
```

```python
report = df.pivot_table(
    index="context_length",
    columns="mode",
    values="peak_memory_mb",
    aggfunc="first"
).reset_index()

report.columns.name = None
report = report.rename(columns={
    "forward": "forward_peak_mb",
    "forward_backward": "training_step_peak_mb"
})

report
```

#### mixed precision 2.7 b memory profiling

```python
import os
import subprocess
import pandas as pd
import re

size = "2.7B"
contexts = [128, 256, 512]
modes = ["forward", "forward_backward"]
precisions = [
    ("fp32", False),
    ("bf16_mixed", True),
]

os.makedirs("memory_snapshots", exist_ok=True)

results = []

for precision_name, use_mixed in precisions:
    for ctx in contexts:
        for mode in modes:
            snapshot_file = f"memory_snapshots/{size}_ctx{ctx}_{mode}_{precision_name}.pickle"

            cmd = f"""
            uv run python ./student/memory_profiling.py \
                --size "{size}" \
                --context_length {ctx} \
                --mode {mode} \
                --warmup_steps 5 \
                --profile_memory \
                --snapshot_file "{snapshot_file}" \
                {"--mixed_precision" if use_mixed else ""}
            """

            print(f"\nRunning size={size} | ctx={ctx} | mode={mode} | precision={precision_name}")

            try:
                output = subprocess.check_output(
                    cmd,
                    shell=True,
                    text=True,
                    stderr=subprocess.STDOUT
                )
                print(output)

                peak_match = re.search(r"Peak memory allocated:\s*([\d\.]+)\s*MB", output)
                param_match = re.search(r"Parameters:\s*([\d\.]+)M", output)

                results.append({
                    "size": size,
                    "context_length": ctx,
                    "mode": mode,
                    "precision": precision_name,
                    "params_m": float(param_match.group(1)) if param_match else None,
                    "peak_memory_mb": float(peak_match.group(1)) if peak_match else None,
                    "snapshot_file": snapshot_file,
                    "status": "ok",
                })

            except subprocess.CalledProcessError as e:
                print(e.output)
                results.append({
                    "size": size,
                    "context_length": ctx,
                    "mode": mode,
                    "precision": precision_name,
                    "params_m": None,
                    "peak_memory_mb": None,
                    "snapshot_file": snapshot_file,
                    "status": "failed/OOM",
                })

df = pd.DataFrame(results)
df
```

```python
comparison = df.pivot_table(
    index=["context_length", "mode"],
    columns="precision",
    values="peak_memory_mb",
    aggfunc="first"
).reset_index()

comparison["absolute_reduction_mb"] = comparison["fp32"] - comparison["bf16_mixed"]
comparison["percent_reduction"] = 100 * (comparison["fp32"] - comparison["bf16_mixed"]) / comparison["fp32"]

comparison
```

## Toy Model

```python
import torch
import torch.nn as nn


class ToyModel(nn.Module):
    def __init__(self, in_features, out_features):
        super().__init__()
        self.fc1 = nn.Linear(in_features, 10, bias=False)
        self.ln = nn.LayerNorm(10)
        self.fc2 = nn.Linear(10, out_features, bias=False)
        self.relu = nn.ReLU()

    def forward(self, x):
        print(f"fc1 weight dtype:     {self.fc1.weight.dtype}")
        x = self.relu(self.fc1(x))
        print(f"fc1 output dtype:     {x.dtype}")
        x = self.ln(x)
        print(f"layer norm output:    {x.dtype}")
        x = self.fc2(x)
        print(f"logits dtype:         {x.dtype}")
        return x


model = ToyModel(20, 5).cuda()
x = torch.randn(4, 20).cuda()

with torch.autocast(device_type="cuda", dtype=torch.float16):
    logits = model(x)
    loss = logits.sum()
    print(f"loss dtype:           {loss.dtype}")

loss.backward()
print(f"fc1 weight grad dtype: {model.fc1.weight.grad.dtype}")
```
